# Question 2

```python
# Declare function
def Bookkeeping():
    # Declare variables
    class_A_cost = 20
    # Ask user for input and assign sold seats
    sold_A_seats = int(input("Please enter how many class A seat you sold: "))
    class_B_cost = 15
    sold_B_seats = int(input("Please enter how many class B seat you sold: "))
    class_C_cost = 10
    sold_C_seats = int(input("Please enter how many class C seat you sold: "))
    # Print tickets sales
    print("Your ticket sales:"
          , "\nClass A seats:", sold_A_seats, ", income:", sold_A_seats * class_A_cost
          , "\nClass B seats:", sold_B_seats, ", income:", sold_B_seats * class_B_cost
          , "\nClass C seats:", sold_C_seats, ", income:", sold_C_seats * class_C_cost
          , "\nTotal sold seats:", sold_A_seats + sold_B_seats + sold_C_seats
          , "\nTotal income:", sold_A_seats * class_A_cost + sold_B_seats * class_B_cost + sold_C_seats * class_C_cost)
    
# Commit function
Bookkeeping()
```
