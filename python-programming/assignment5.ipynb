```python
class Book:
    def __init__(self, title, author, publication_year):
        self.title = title
        self.author = author
        self.publication_year = publication_year

    def __eq__(self, other):
        return self.title == other.title and self.author == other.author and self.publication_year == other.publication_year
    
    def __hash__(self):
        return hash((self.title, self.author, self.publication_year))

BoolSet = set()

def add_book(title, author, publication_year):
    book = Book(title, author, publication_year)
    BoolSet.add(book)

def view_books():
    for book in BoolSet:
        print(f"{book.title} by {book.author} ({book.publication_year})")

def search_books(search_term):
    if search_term in BoolSet:
        print(f"{search_term.title} by {search_term.author} ({search_term.publication_year})")
    else:
        print("Book " + search_term + " not found")

def user_input():
    while (user_option := input("Enter a to add a book, v to view all books, s to search for a book, or q to quit: ")) != 'q':
        if user_option == 'a':
            title = input("Enter the title of the book: ")
            author = input("Enter the author of the book: ")
            publication_year = input("Enter the publication year of the book: ")
            add_book(title, author, publication_year)
        elif user_option == 'v':
            view_books()
        elif user_option == 's':
            search_term = input("Enter the title of the book you are searching for: ")
            search_books(search_term)
        else:
            print("Invalid option")

user_input()
```
