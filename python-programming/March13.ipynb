```python
text = "Hello, Python!!!"
clean_text = text.rstrip("!")
 
print("Before rstrip:", repr(text))
print("After rstrip:", repr(clean_text))
```

```python
f = open(".file", "w")
f.write("Hello, World!")
f.close()
f = open(".file", "a")
f.write("Hello, Python!!!")
f.close()
f = open(".file", "r")
text = f.read()
print(text)
f.close()
```

```python
try:
    f = open(".file", "r")
    text = f.read()
    print(text)
finally:
    f.close()
```

```python
with open(".file", "r") as f:
    for line in f:
        print(line)
```

```python
with open(".file", "r") as f:
    for line in f:
        print(line.rstrip())
```

```python
with open(".file", "r") as f:
    count = 1
    for line in f:
        print("line ", count, ": ", line.strip())
        count += 1
```
